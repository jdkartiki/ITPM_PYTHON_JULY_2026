```python
# continue 

#skip 5 number 

for i in range(1,10):
    if i==5:
        continue
    print(i)
```

```python
# skip even numbers from 1 to 10

for i in range(1,11):
    if i%2==0:
        continue
    print(i)
```

```python
# skip odd numbers from 1 to 10

for i in range(1,11):
    if i%2!=0:
        continue
    print(i)
```

```python
# skip negative numbers
nums=[1,0,-2,-3,-4,-5,-6,10,20]

for i in nums:
    if i<0:
        continue
    print(i)
```

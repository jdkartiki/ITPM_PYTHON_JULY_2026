```python
# print sqaure pattern of 5*5 

for i in range(5):
    for j in range(5):
        print("*",end=" ")
    print()
```

```python
# right angles triangle 

for i in range(1,6):
    for j in range(i):
        print("*",end=" ")
    print()
```

```python
# right angles triangle using column num
for i in range(1,6):
    for j in range(1,i+1):
        print(j,end=" ")
    print()
```

```python
# right angles triangle using row num
for i in range(1,6):
    for j in range(1,i+1):
        print(i,end=" ")
    print()
```

```python
# reverse pattern

for i in range(5,0,-1):
    for j in range(i):
        print("*",end=" ")
    print()
```

```python
# reverse pattern using column num

for i in range(5,0,-1):
    for j in range(1,i+1):
        print(j,end=" ")
    print()
```

```python
# reverse pattern using row num

for i in range(5,0,-1):
    for j in range(1,i+1):
        print(i,end=" ")
    print()
```

```python
# characters

for i in range(1,6):
    for j in range(i):
        print(chr(64+i),end=" ")
    print()
```

```python

num=1
for i in range(1,7):
    for j in range(i):
        print(num,end=" ")
        num=num+1
    print()
```

```python
# subjects allocated to students

students=["Ram","Raj","Jay"]

for i in students:
    print("Student name is :: ",i)

    for subject in ["Java","Python","SQL"]:
        print("Subject is :: ",subject)

    print()
```

```python
# cinema seating arrangement

for row in range(1,4):
    print("Row number :: ",row)

    for seats in range(1,6):
        print("Seat no. :: ",seats,end="\t")

    print("\n")
```

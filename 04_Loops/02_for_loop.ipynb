#### range() :
* range() is used to give a particular range of number 
* it is an inbuilt function
* range(start number, ending number+1)
* range() will take ending number only upto given number 
* eg :: range(1,10)--- 9 times    range(1,11) -- 10 times 

```python
# print 1 to 10 
# for(i=1;i<=10;i++)

for i in range(1,11):
    print(i)
```

```python
# accept num and calculate the table for it 

num=int(input("Enter number :"))

for i in range(1,11):
    print(num," * ",i," = ",num*i)
```

```python
# accept the num and calculate the factorial of it 

num=int(input("Enter number :: "))
fact=1

for i in range(1,num+1):
    fact=fact*i


print("Factorial is :: ",fact)
```

#### A company has 5 employees using for loop, take each employee's salary and calculate
* total salary
*  Average Salary
* number of emp earning more than 50K

```python
total_sal=0
emp_count_above=0

for i in range(1,6):
    salary=float(input(f"Enter Salary of employee {i} :: Rs. "))

    total_sal=total_sal+salary

    if salary>50000:
        emp_count_above+=1


average_sal=total_sal/5

print("======== Salary Report ============")
print("Total salary :: ",total_sal)
print("Average Salary :: ",average_sal)
print("Employees earning more than Rs. 50K :: ",emp_count_above)
```

### A customer buys 5 products . take the name and price of product  and calculate the total bill

A customer purchase 5 diffferent procucts 
* take price of each product
* calculate total bill
* if total is Rs. 5000 or more give 10% discount 
* otherwise give 5% discount
* display the final payable amount

```python
total=0

for i in range(1,6):
    price=float(input(f"Enter price of product {i} :: Rs. "))
    total=total+price

if total>5000:
    discount=total*0.10
else:
    discount=total*0.05

final_amount=total-discount

print("========== SHOPPING BILL =============")
print("Total Bill :: Rs. ",total)
print("Discount amount :: Rs. ",discount)
print("Total Payable Bill :: Rs. ",final_amount)
```

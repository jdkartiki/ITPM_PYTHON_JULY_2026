```python
# ATM MACHINE CODE

choice=int(input("Enter Your Choice:: \n1.Balance Enquiry \n2.Deposit Money \n3.Withdraw Money"))

match choice:
    case 1:
        print("Balance Check")
    case 2:
        print("Deposit Money")
    case 3:
        print("Withdraw Money")
    case _:
        print("Invalid Choice")
```

```python
# check traffic signal 

signal=input("Enter colour of signal :: ")

match signal:
    case "red":
        print("STOP")
    case "green":
        print("GO")
    case "yellow":
        print("SLOW ")
    case _:
        print("Invalid Signal Input")
```

```python
# emp role 

emp_role=input("Enter role of employee :: ")

match emp_role:
    case "manager":
        print("Employee is Manager")
    case "developer":
        print("Employee is Developer")
    case "tester":
        print("Employee is Tester")
    case "HR":
        print("Employee is HR")
    case _:
        print("Invalid Employee Role")
```

```python
# food ordering system

itemname=input("Enter Food NAme :: [pizza/burger/sandwitch]")

match itemname:
    case "pizza":
        print("Price is :: Rs. 500")
    case "burger":
        print("Price is :: Rs. 400")
    case "sandwitch":
        print("Price is :: Rs. 300")
    case _:
        print("Please select some Food from Menu")


```

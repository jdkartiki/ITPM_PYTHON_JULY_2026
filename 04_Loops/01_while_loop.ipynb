* Daily Steps
    * Keep entering the number of steps walked each day.
    * Stop when 0 is entered.
    * Display the total steps.


* Pocket Money
    * Keep entering the amount of pocket money received.
    * Stop when 0 is entered.
    * Display the total money.


* Book Reading
    * Enter the number of pages read each day.
    * Stop when 0 is entered.
    * Display the total pages read.


* Class Attendance
    * Enter P for Present or A for Absent.
    * Enter Stop to finish.
    * Count the number of Present students.

## 1. Daily Steps

```python
total_steps = 0

while True:
    steps = int(input("Enter steps walked (0 to stop): "))

    if steps == 0:
        break

    total_steps += steps

print("Total Steps Walked:", total_steps)
```

---

## 2. Pocket Money

```python
total_money = 0

while True:
    money = int(input("Enter pocket money received (0 to stop): "))

    if money == 0:
        break

    total_money += money

print("Total Pocket Money Received: ₹", total_money)
```

## 3. Book Reading

```python
total_pages = 0

while True:

    if pages == 0:
        break

    total_pages += pages

print("Total Pages Read:", total_pages)
```

---

## 4. Class Attendance

```python
present = 0

while True:
    attendance = input("Enter P for Present, A for Absent (Stop to finish): ")

    if attendance.lower() == "stop":
        break

    if attendance.upper() == "P":
        present += 1

print("Total Present Students:", present)
```

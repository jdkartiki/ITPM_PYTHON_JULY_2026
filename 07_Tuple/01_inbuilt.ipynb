# 📘 Python Tuples - Complete Notes

---

## 1. What is a Tuple?

A **tuple** in Python is a collection of ordered elements enclosed in parentheses `( )`.

* Tuples can store multiple data types
* Tuples are **immutable** (cannot be changed after creation)

### Example

```python
numbers = (10, 20, 30, 40)
mixed = (1, "Python", 3.5, True)
```

---

## 2. Characteristics of Tuples

* ✔ Ordered (maintains insertion order)
* ❌ Immutable (cannot modify elements)
* ✔ Allows duplicate values
* ✔ Supports indexing & slicing
* ✔ Can store different data types
* ✔ Faster than lists (performance advantage)

---

## 3. Creating Tuples

### Empty Tuple

```python
t = ()
```

### Without Parentheses

```python
t = 1, 2, 3
```

### Single Element Tuple (Important)

```python
t = (5,)   # comma is required
```

### Using tuple() Constructor

```python
t = tuple([1, 2, 3])
```

### Nested Tuple

```python
t = ((1, 2), (3, 4))
```

---

## 4. Tuple Indexing

### Positive Indexing

```python
t = (10, 20, 30, 40)
print(t[0])   # 10
print(t[2])   # 30
```

### Negative Indexing

```python
print(t[-1])  # 40
print(t[-3])  # 20
```

### Indexing Diagram

```
Index:   0    1    2    3
Value:  10   20   30   40
Neg:    -4   -3   -2   -1
```

---

## 5. Tuple Slicing

### Syntax

```python
tuple[start : end : step]
```

### Examples

```python
t = (10, 20, 30, 40, 50)

print(t[1:4])   # (20, 30, 40)
print(t[:3])    # (10, 20, 30)
print(t[::2])   # (10, 30, 50)
print(t[::-1])  # Reverse tuple
```

---

## 6. Immutability of Tuples

### Attempt to Modify (Error)

```python
t = (10, 20, 30)
t[1] = 99   # ❌ TypeError
```

### Workaround (Convert to List)

```python
t = (10, 20, 30)
lst = list(t)
lst[1] = 99
t = tuple(lst)
print(t)
```

## 7. Common Tuple Methods

| Method  | Description              |
| ------- | ------------------------ |
| count() | Count occurrences        |
| index() | Find position of element |

### Example

```python
t = (1, 2, 2, 3)
print(t.count(2))   # 2
print(t.index(3))   # 3
```

## 8. Traversing Tuples

### Using for Loop

```python
t = (10, 20, 30)
for i in t:
    print(i)
```

### Using while Loop

```python
t = (10, 20, 30)
i = 0

while i < len(t):
    print(t[i])
    i += 1
```

---

## 9. Tuple Operators

### Concatenation

```python
a = (1, 2)
b = (3, 4)
print(a + b)
```

### Repetition

```python
print((1, 2) * 3)
```

### Membership

```python
print(20 in (10, 20, 30))
```

---

## 10. Tuple Packing & Unpacking

### Packing

```python
t = 1, 2, 3
```

### Unpacking

```python
a, b, c = t
print(a, b, c)
```

### Extended Unpacking

```python
a, *b = (1, 2, 3, 4)
print(a)  # 1
print(b)  # [2, 3, 4]
```

---

## 11. Nested Tuples

### Example

```python
t = (
    (1, 2, 3),
    (4, 5, 6)
)

for row in t:
    for val in row:
        print(val, end=" ")
```

---

## 12. Copying Tuples

### Direct Assignment (Safe)

```python
t1 = (1, 2, 3)
t2 = t1
```

*(Since tuples are immutable, this is safe)*

# 📘 Python Functions - Complete Notes

---

## 1. What is a Function?

A **function** is a block of reusable code that performs a specific task.

* Helps in code reusability
* Makes code modular and readable

### Syntax

```python
def function_name(parameters):
    # code block
    return value
```

### Example

```python
def greet():
    print("Hello, Welcome!")

greet()
```

---

## 2. Types of Functions in Python

### 1. Built-in Functions

Predefined functions available in Python.

```python
print(len("Python"))   # 6
print(max(10, 20, 30))  # 30
```

---

### 2. User-defined Functions

Functions created by the user using `def`.

```python
def add(a, b):
    return a + b

print(add(2, 3))
```

---

### 3. Anonymous Functions (Lambda)

Functions without a name, defined using `lambda`.

```python
square = lambda x: x * x
print(square(5))
```

---

### 4. Recursive Functions

Function that calls itself.

```python
def factorial(n):
    if n == 1:
        return 1
    return n * factorial(n-1)

print(factorial(5))
```

---

## 3. Types of User-Defined Functions (Based on Arguments & Return)

### 1. No Arguments, No Return Value

```python
def greet():
    print("Hello")

greet()
```

---

### 2. Arguments, No Return Value

```python
def greet(name):
    print("Hello", name)

greet("Kartiki")
```

---

### 3. No Arguments, Return Value

```python
def get_number():
    return 10

print(get_number())
```

---

### 4. Arguments with Return Value

```python
def add(a, b):
    return a + b

print(add(5, 3))
```

---

## 4. Types of Function Parameters

### 1. Positional Arguments

Arguments passed in correct order.

```python
def add(a, b):
    print(a + b)

add(5, 3)
```

---

### 2. Keyword Arguments

Arguments passed using parameter names.

```python
def student(name, age):
    print(name, age)

student(age=20, name="Kartiki")
```

---

### 3. Default Arguments

Default values assigned to parameters.

```python
def greet(name="Guest"):
    print("Hello", name)

greet()
greet("Amit")
```

---

### 4. Variable Length Arguments (*args)

Used when number of arguments is unknown.

```python
def total(*numbers):
    print(sum(numbers))

total(1, 2, 3, 4)
```

---

### 5. Keyword Variable Length Arguments (**kwargs)

Accepts keyword arguments.

```python
def details(**data):
    print(data)

details(name="Kartiki", age=30)
```

---

## 5. Return Statement

* Used to return value from function
* A function can return multiple values

```python
def calc(a, b):
    return a+b, a-b

x, y = calc(5, 3)
print(x, y)
```

---

## 6. Scope of Variables

### Local Scope

Variable inside function

```python
def test():
    x = 10
    print(x)

test()
```

---

### Global Scope

Variable outside function

```python
x = 20

def test():
    print(x)

test()
```

---

### Global Keyword

```python
x = 10

def change():
    global x
    x = 50

change()
print(x)
```

---

## 7. Nested Functions

Function inside another function.

```python
def outer():
    def inner():
        print("Inside inner")
    inner()

outer()
```

---

## 8. Important Function Programs (Exam)

### Find Maximum of Two Numbers

```python
def maximum(a, b):
    return a if a > b else b

print(maximum(10, 20))
```

---

### Check Even or Odd

```python
def check(n):
    return "Even" if n % 2 == 0 else "Odd"

print(check(5))
```

---

### Sum of List using Function

```python
def total(lst):
    return sum(lst)

print(total([1, 2, 3]))
```

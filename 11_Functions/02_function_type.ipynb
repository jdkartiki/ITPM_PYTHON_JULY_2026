Types of IMPLEMENTATION OF Function 
1. NRNP : NO RETURN NO PARAMETER
2. NRWP : NO RETURN WITH PARAMETER
3. RNP: RETURN WITH NO PARAMETER
4. RP: RETURN WITH PARAMETER

```python
# function without argument and without return type
# NRNP

def greet():
    print("Hello , Welcome to Function Lecture ")

greet()
greet()
```

```python
#  function without argument and without return type
# NRNP

def add():
    a=int(input("Enter Value of a :: "))
    b=int(input("Enter Value of b :: "))
    sum=a+b
    print("Sum is :: ",sum)

add()
add()
add()
```

```python
#  function without argument and without return type
# NRNP
def print_num():
    for i in range(1,11):
        print(i)


print_num()
```

```python
# no argument but with return type
# WRNP

def welcome():
    return "Welcome to the Event"

welcome()
```

```python
def add():
    return 10+20+30

add() 
```

```python
def multiply():
    num=int(input("Enter num :: "))
    return num*2

multiply()
# print("Output is :: ",multiply())
```

```python
#NPWR
def elemets():
    return [10,20,30,40]

elemets()
```

```python
def even_odd():
    num=int(input("Enter num :: "))
    if num%2==0:
        return "Even number"
    else:
        return "Odd Number"

even_odd()
    
```

```python
# with argument , without return
# NRWP

def add(a,b):
    print("Sum is :",a+b)

add(5,10)
add(10,20)
add(50,100)
```

```python
# user input

def area(r):
    area=3.14*r*r
    print("Area of Circle is :: ",area)


r=float(input("enter radius :: "))
area(r)
```

```python
# even number 

def even_num(n):
    for i in range(2,n+1,2):
        print(i)

even_num(20)
```

```python
# with arg with parameter
# WRWP

def factorial(n):
    fact=1
    for i in range(1,n+1):
        fact=fact*i
    return fact


n=int(input("Enter number :: "))
print("Factorial of num is :: ",factorial(n))
```

```python
def max_num(a,b):
    if a>b:
        return a
    return b

print(max_num(10,20))
```

# Python Environment Setup (Windows & macOS)

---

## 📌 1. What is Python Environment Setup?

Setting up a Python environment means installing Python and configuring tools required to write, run, and manage Python programs efficiently.

It includes:

* Installing Python
* Setting environment variables
* Installing IDE (VS Code / Jupyter Notebook)
* Managing packages using pip
* Creating virtual environments

---

## 🖥️ 2. Python Setup on Windows

### ✅ Step 1: Download Python

* Go to official website: https://www.python.org
* Click on **Downloads → Windows**
* Download latest Python version

---

### ✅ Step 2: Install Python

1. Run the downloaded `.exe` file
2. ✔️ IMPORTANT: Check

   ```
   Add Python to PATH
   ```
3. Click **Install Now**

---

### ✅ Step 3: Verify Installation

Open Command Prompt and run:

```bash
python --version
```

or

```bash
python
```

If installed correctly, Python version will be displayed.

---

### ✅ Step 4: Install pip (Package Manager)

pip is usually installed by default.

Check:

```bash
pip --version
```

---

### ✅ Step 5: Install VS Code

* Download from: https://code.visualstudio.com
* Install VS Code

---

### ✅ Step 6: Install Python Extension in VS Code

1. Open VS Code
2. Go to Extensions (`Ctrl + Shift + X`)
3. Search: **Python (by Microsoft)**
4. Click **Install**

---

### ✅ Step 7: Install Jupyter Notebook

```bash
pip install notebook
```

Run Jupyter:

```bash
jupyter notebook
```

---

##  3. Python Setup on macOS

### ✅ Step 1: Check Python Pre-installed

Open Terminal:

```bash
python3 --version
```

---

### ✅ Step 2: Install Python (if not installed)

* Download from: https://www.python.org
  OR use Homebrew:

```bash
brew install python
```

---

### ✅ Step 3: Verify Installation

```bash
python3 --version
```

---

### ✅ Step 4: Install pip

```bash
pip3 --version
```

---

### ✅ Step 5: Install VS Code

```bash
brew install --cask visual-studio-code
```

---

### ✅ Step 6: Install Python Extension

Same steps as Windows:

* Open VS Code
* Install Python extension

---

### ✅ Step 7: Install Jupyter Notebook

```bash
pip3 install notebook
```

Run:

```bash
jupyter notebook
```

---

## 📦 4. Virtual Environment Setup

### 🔹 Why Use Virtual Environment?

* Isolate project dependencies
* Avoid version conflicts

---

### 🔹 Create Virtual Environment

**Windows:**

```bash
python -m venv myenv
```

**macOS:**

```bash
python3 -m venv myenv
```

---

### 🔹 Activate Virtual Environment

**Windows:**

```bash
myenv\Scripts\activate
```

**macOS:**

```bash
source myenv/bin/activate
```

---

### 🔹 Deactivate Environment

```bash
deactivate
```

---

## 📥 5. Installing Packages

```bash
pip install package_name
```

Example:

```bash
pip install numpy pandas matplotlib
```

---

## 🧪 6. Running Python Programs

### 🔹 Using Command Line

```bash
python file.py
```

or (macOS):

```bash
python3 file.py
```

---

### 🔹 Using VS Code

1. Open file `.py`
2. Click **Run ▶️**
3. Or use:

```
Ctrl + F5
```

---

## 📒 7. Using Jupyter Notebook

Start:

```bash
jupyter notebook
```

Features:

* Interactive coding
* Supports Markdown + Code

## ⚠️ 8. Common Errors & Solutions

| Error                   | Solution                                |
| ----------------------- | --------------------------------------- |
| Python not recognized   | Add Python to PATH                      |
| pip not found           | Reinstall Python / ensure pip installed |
| Permission denied (Mac) | Use `sudo` or fix permissions           |
| Module not found        | Install using pip                       |

---

* Install Python from official website
* Set PATH correctly

* Use pip for package management
* Use virtual environments for projects

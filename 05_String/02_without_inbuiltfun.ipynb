```python
# reverse a string 
str=input("Enter String :: ")
rev=""

for ch in str:
    rev=ch+rev 

print("Reveresed String :: ",rev)
```

```python
# find occurance of character
str=input("Enter String :: ")
max_char=""
max_count=0

# is for iterating the str for count 
for ch in str:
    count=0
    # is used to check first character with second
    for c in str:
        if ch==c:
            count=count+1

        if count>max_count:
            max_count=count
            max_char=ch

print("Maximum Occuring Character is :: ",max_char)
print("Maximum  Character Count is :: ",max_count)
```

```python
# count the occurance of given character
str=input("Enter String :: ")
ch=input("Enter character to find ocuurance :: ")
count=0

for c in str.lower():
    if c==ch:
        count+=1

print("Occurance of given character is :: ",count)
```

```python
# remove the extra spaces from string
# strip - remove spaces from start and end 

str=input("Enter String :: ")
res=""

for ch in str:
    if ch!=" ":
        res+=ch


print("String without Spaces :: ",res)
```

```python
# copy one string to another

str=input("Enter String :: ")
res=""

for ch in str:
    res+=ch

print("Orginal string :: ",str)
print("Copied string :: ",res)
```

```python
# remove the duplicates
str=input("Enter String :: ")
res=""

for ch in str:
    if ch not in res:
        res+=ch

print("String without duplication is :: ",res)
```

```python
# compare two strings

s1=input("Enter String 1 :: ")
s2=input("Enter String 2 :: ")

if len(s1)!=len(s2):
    print("Strings are not equal ")
else:
    equal=True
    for i in range(len(s1)):
        if s1[i]!=s2[i]:
            equal=False
            break

    if equal:
        print("String are equal")
    else:
        print("String are not Equal")

```

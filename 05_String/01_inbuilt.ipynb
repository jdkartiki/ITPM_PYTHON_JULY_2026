```python
# string - uppercase,lowercase,titlecase,capitalize

str=input("Enter String :: ")
print("String in Lowercase :: ",str.lower())
print("String in Uppercase :: ",str.upper())
print("String in Titlecase :: ",str.title())
print("String in Capitalize Case :: ",str.capitalize())
```

```python
# count the occurrence of specific word 
str=input("Enter String :: ")
word=input("Enter word to calculate occurance :: ")

print("String is :: ",str)
print("Count of word in given string is :: ",str.count(word))
```

```python
# find specific word 
str=input("Enter String :: ")
word=input("Enter word to search :: ")

print("String is :: ",str)
print("Count of word in given string is :: ",str.find(word))
```

```python
# join the string
words=["India","is","my","country"]

res=" ".join(words)
print("String is :: ",res)

res="-".join(words)
print("String is :: ",res)

res=":".join(words)
print("String is :: ",res)

res=",".join(words)
print("String is :: ",res)

res="|".join(words)
print("String is :: ",res)

res="=".join(words)
print("String is :: ",res)
```

```python
# check whether my string contains only digit 

str=input("Enter String :: ")

if str.isdigit():
    print("String contains only Digit")
else:
    print("String contains both")
```

```python
# check whether my string contains only alpahbets 

str=input("Enter String :: ")

if str.isalpha():
    print("String contains only alphabets")
else:
    print("String contains both")
```

```python
# check whether my string contains only both 

str=input("Enter String :: ")

if str.isalnum():
    print("String contains both alphabets and digits")
else:
    print("String contains either alphabets or digits")
```

```python
# starts with and ends with 

str=input("Enter String :: ")
sword=input("Enter Word to check the starting word :: ")

if str.startswith(sword):
    print("Yes String starts with given word ")
else:
    print("No String don't starts with given word ")

eword=input("Enter Word to check the ending word :: ")

if str.endswith(eword):
    print("Yes String ends with given word ")
else:
    print("No String don't ends with given word ")
```

```python
# check count of vowels in string

str=input("Enter String :: ")
count=0

for ch in str.lower():
    if ch in "aeiou":
        count=count+1

print("Total Vowel COunt is :: ",count)
```

```python
# reverse a string

str=input("Enter String :: ")
print("Reversed String id :: ",str[::-1])
```

```python
# string is palindrome
str=input("Enter String :: ")

if str==str[::-1]:
    print("Palindrome String")
else:
    print("NOT a Palindrome string")
```

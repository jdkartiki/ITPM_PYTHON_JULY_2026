Dunder Methods in Python (Magic Methods)

**Dunder = “Double UNDERscore” methods**
They are special methods with names like:

```python
__init__, __str__, __len__, __add__
```

Called **automatically by Python** to perform built-in operations.

---

What are Dunder Methods?

Special methods that let you **define how your objects behave**

Example:

```python
print(obj)     → calls obj.__str__()
len(obj)       → calls obj.__len__()
a + b          → calls a.__add__(b)
```

---

Common Dunder Methods

| Method     | Use                               |
| ---------- | --------------------------------- |
| `__init__` | Constructor (initialize object)   |
| `__str__`  | String representation (for print) |
| `__repr__` | Developer representation          |
| `__len__`  | Length using `len()`              |
| `__add__`  | `+` operator                      |
| `__sub__`  | `-` operator                      |
| `__eq__`   | `==` comparison                   |

---

Example 1: `__init__` (Constructor)

```python
class Student:
    def __init__(self, name):
        self.name = name

s = Student("Kartiki")
print(s.name)
```

Runs automatically when object is created

---

Example 2: `__str__`

```python
class Student:
    def __init__(self, name):
        self.name = name

    def __str__(self):
        return f"Student name is {self.name}"

s = Student("Kartiki")
print(s)
```

Output:

```
Student name is Kartiki
```

---

Example 3: `__len__`

```python
class MyList:
    def __init__(self, items):
        self.items = items

    def __len__(self):
        return len(self.items)

obj = MyList([1, 2, 3, 4])
print(len(obj))   # Output: 4
```

---

Example 4: `__add__`

```python
class Number:
    def __init__(self, value):
        self.value = value

    def __add__(self, other):
        return self.value + other.value

a = Number(10)
b = Number(20)

print(a + b)   # Output: 30
```

---

Why Use Dunder Methods?

* Make objects behave like **built-in types**
* Enable **operator overloading**
* Improve **readability & usability**

---

Real-Life Meaning

* `__init__` → “When object is created, do this”
* `__str__` → “How to display object”
* `__add__` → “What happens when we use +”

---

* Important Note

👉 Don’t call them directly:

```python
obj.__str__()   ❌ (avoid)
print(obj)      ✅ (correct)
```

````python
class Bank:
    def __init__(self,name,balance):
        self.name=name
        self.balance=balance

    def deposit(self,damt):
        self.balance+=damt
        print("Deposited AMount is :: ",damt)
        print("Balance AMount is :: ",self.balance)

    def withdraw(self,wamt):
        self.balance-=wamt
        print("Withdraw AMount is :: ",wamt)
        print("Balance AMount is :: ",self.balance)


name=input("Enter Account Holder Name :: ")
balance=float(input("Enter Balance :: "))

acc=Bank(name,balance)

damt=float(input("Enter Deposit AMount :: "))
acc.deposit(damt)

wamt=float(input("Enter Withdraw AMount :: "))
acc.withdraw(wamt)
    
````

````python
# Product Details 
class Product:
    def __init__(self,name,price,quantity):
        self.name=name
        self.price=price
        self.quantity=quantity


    def total(self):
        return self.price*self.quantity
    

name=input("Enter Name :: ")
price=float(input("Enter Price :: "))
quantity=int(input("enter Quantity :: "))

p=Product(name,price,quantity)
print("Product Name is :: ",p.name)
print("Price for Each Product is :: ",p.price)
print("Quantity of Product is :: ",p.quantity)
print("Total Cost is :: ",p.total())
````

### Problem Statement: Car Rental System

Create a **Python program using a class and object** to develop a simple **Car Rental System**.

The program should accept the following details from the user:

* **Car Name**
* **Rent Per Day**
* **Number of Days**

Create a class named `CarRental` that contains:

1. A constructor to initialize the car details.
2. A method to calculate the **total rent**.
3. A method to display all rental details.

The total rent should be calculated using:

```text
Total Rent = Rent Per Day × Number of Days
```

### Sample Input

```text
Enter car name: Swift
Enter rent per day: 1500
Enter number of days: 4
```

### Expected Output

```text
===== CAR RENTAL DETAILS =====
Car Name: Swift
Rent Per Day: 1500
Number of Days: 4
Total Rent: 6000
```

**Requirement:** Use **class, constructor, instance variables, methods, and object creation** to solve the problem.


````python
class Car:

    def __init__(self,car_name,rent_per_day,total_days):
        self.car_name=car_name
        self.rent_per_day=rent_per_day
        self.total_days=total_days


    def total_rent(self):
        return self.rent_per_day*self.total_days

    def display(self):
        print("============= CAR RENTAL SYSTEM =======")
        print("Car NAme :: ",self.car_name)
        print("Rent Per Day :: ",self.rent_per_day)
        print("Total Days :: ",self.total_days)
        print("Total Rent :: ",self.total_rent())



car_name=input("Enter Car NAme :: ")
rent_per_day=float(input("Enter rent per Day :: "))
total_days=int(input("Enter Total Days :: "))

car=Car(car_name,rent_per_day,total_days)

car.display()
````

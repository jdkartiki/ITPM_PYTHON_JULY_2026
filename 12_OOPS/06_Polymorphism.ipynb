````python
## 🔷 Polymorphism in Python

**Polymorphism** means *“many forms”*. In Python, it allows the same function or method name to behave differently depending on the object or data type.

👉 In simple words:
**One interface → Multiple implementations**

---

# 🔹 Types of Polymorphism in Python

## 1. Method Overriding (Runtime Polymorphism)

When a **child class provides a different implementation** of a method that already exists in the parent class.

### ✅ Example:

```python
class Animal:
    def speak(self):
        print("Animal makes sound")

class Dog(Animal):
    def speak(self):
        print("Dog barks")

class Cat(Animal):
    def speak(self):
        print("Cat meows")


# Usage
a = Animal()
d = Dog()
c = Cat()

a.speak()
d.speak()
c.speak()
```

### 🔍 Output:

```
Animal makes sound
Dog barks
Cat meows
```

👉 Same method `speak()` behaves differently.

---

## 2. Method Overloading (Compile-time style)

Python does **not support true method overloading**, but we can achieve it using:

* Default arguments
* `*args`

### ✅ Example using default arguments:

```python
class Math:
    def add(self, a, b, c=0):
        return a + b + c

m = Math()

print(m.add(2, 3))       # 5
print(m.add(2, 3, 4))    # 9
```

👉 Same method works with different number of arguments.

---

## 3. Operator Overloading

Python allows operators (`+`, `-`, `*`) to work differently for objects using **magic methods**.

### ✅ Example:

```python
class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __add__(self, other):
        return Point(self.x + other.x, self.y + other.y)

    def display(self):
        print(self.x, self.y)


p1 = Point(2, 3)
p2 = Point(4, 5)

p3 = p1 + p2
p3.display()
```

### 🔍 Output:

```
6 8
```

👉 `+` operator works for custom objects.

---

## 4. Duck Typing (Dynamic Polymorphism)

Python focuses on **behavior, not type**.

👉 “If it looks like a duck and behaves like a duck, it is a duck.”

### ✅ Example:

```python
class Bird:
    def fly(self):
        print("Bird can fly")

class Airplane:
    def fly(self):
        print("Airplane flies")

def make_fly(obj):
    obj.fly()

b = Bird()
a = Airplane()

make_fly(b)
make_fly(a)
```

### 🔍 Output:

```
Bird can fly
Airplane flies
```

👉 No inheritance required — just same method name.

---

# 🔥 Summary Table

| Type                 | Description                                    |
| -------------------- | ---------------------------------------------- |
| Method Overriding    | Same method, different behavior in child class |
| Method Overloading   | Same method, different parameters              |
| Operator Overloading | Operators behave differently                   |
| Duck Typing          | Object behavior matters, not type              |

---

# 🚀 Real-Life Example

Think of a **remote control**:

* Same button → different TVs react differently
  👉 That’s polymorphism!

---

If you want, I can give:

* 💯 Interview questions on polymorphism
* 🧠 Tricky coding problems
* 📊 Diagrams for better understanding

Just tell me 👍
````

##### Explain Compile-time Polymorphism and Run-time Polymorphism with Examples

---

##### 1. Compile-time Polymorphism (Static Polymorphism)

##### Definition

Compile-time polymorphism is the type of polymorphism in which the method to be executed is determined **during compilation**.

In Python, true compile-time polymorphism is **not supported** because Python is a dynamically typed language. However, it can be **simulated using default arguments** (or variable-length arguments).

##### Advantages

- Faster execution.
- Improves code readability.
- Allows the same method to work with different numbers of arguments.

##### Example (Method Overloading using Default Arguments)

```python
class Calculator:

    def add(self, a, b, c=0):
        print("Addition =", a + b + c)

c = Calculator()

c.add(10, 20)        # Two arguments
c.add(10, 20, 30)    # Three arguments
```

##### Output

```text
Addition = 30
Addition = 60
```

##### Explanation

- The same method `add()` performs different operations based on the number of arguments passed.
- This is known as **method overloading**, which represents **compile-time polymorphism** in Python.

---

##### 2. Run-time Polymorphism (Dynamic Polymorphism)

###### Definition

Run-time polymorphism is the type of polymorphism in which the method to be executed is decided **while the program is running**.

In Python, this is achieved using **Method Overriding**.

##### Advantages

- Supports flexibility.
- Allows child classes to provide their own implementation.
- Improves code reusability.

##### Example (Method Overriding)

```python
class Animal:

    def sound(self):
        print("Animal makes a sound")

class Dog(Animal):

    def sound(self):
        print("Dog barks")

obj = Dog()
obj.sound()
```

##### Output

```text
Dog barks
```

##### Explanation

- The parent class `Animal` has a `sound()` method.
- The child class `Dog` overrides the same method.
- When the object of `Dog` calls `sound()`, Python executes the child class method.
- This decision is made **at runtime**, so it is called **Run-time Polymorphism**.

---

##### Difference Between Compile-time and Run-time Polymorphism

##### Difference Between Compile-time Polymorphism and Run-time Polymorphism

### 1. Decision Time
- **Compile-time Polymorphism:** The method to be executed is decided during **compilation**.
- **Run-time Polymorphism:** The method to be executed is decided during **program execution (runtime)**.

---

### 2. Technique Used
- **Compile-time Polymorphism:** Achieved using **Method Overloading** (simulated using default arguments in Python).
- **Run-time Polymorphism:** Achieved using **Method Overriding**.

---

### 3. Speed
- **Compile-time Polymorphism:** Faster because the method is selected before execution.
- **Run-time Polymorphism:** Slightly slower because the method is selected while the program is running.

---

### 4. Method Behavior
- **Compile-time Polymorphism:** The same method performs different tasks based on the **number or type of arguments**.
- **Run-time Polymorphism:** The **child class provides its own implementation** of the parent class method.

---

### 5. Binding
- **Compile-time Polymorphism:** Uses **Early Binding (Static Binding)**.
- **Run-time Polymorphism:** Uses **Late Binding (Dynamic Binding)**.

---

### 6. Python Support
- **Compile-time Polymorphism:** Python does not support true compile-time polymorphism, but it can be simulated using **default arguments**.
- **Run-time Polymorphism:** Fully supported in Python through **Method Overriding**.

---

 🦆 Duck Typing (in detail)

**Duck typing** is a concept in dynamic programming languages (like Python) where the **type or class of an object is less important than the methods or behaviors it defines**.

The name comes from the famous phrase:

> *“If it walks like a duck and quacks like a duck, then it must be a duck.”*

In programming terms:

* You don’t check an object’s type explicitly.
* You just **use the object**, assuming it has the required methods/attributes.
* If it behaves correctly → it works.
* If not → you get a runtime error.

---

 🔍 Key Characteristics

* No need for inheritance or interfaces.
* Focus on **behavior, not type**.
* Common in dynamically typed languages like Python.
* Promotes flexibility and polymorphism.

---

 Simple Example

```python
class Duck:
    def speak(self):
        print("Quack!")

class Dog:
    def speak(self):
        print("Bark!")

def make_sound(animal):
    animal.speak()

make_sound(Duck())
make_sound(Dog())
```

👉 Both objects work because they have a `speak()` method — **not because they share a parent class**.

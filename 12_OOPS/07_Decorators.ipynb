Decorators in Python (Brief Explanation)

A decorator in Python is a function that modifies or extends the behavior of another function without changing its actual code.

👉 Think of it like adding extra features (logging, validation, timing, etc.) to a function.


What is a Decorator? (Detailed Theory)

A **decorator** is a **function that takes another function as input, adds some functionality, and returns a new function**.

👉 In simple words:
**Decorator = Wrapper around a function**

---

🔹 Why Do We Use Decorators?

Without decorators, you would have to repeat code like:

```python
print("Logging started")
function()
print("Logging ended")
```

👉 If many functions need this → code repetition ❌

✔ Decorators solve this by **reusing common logic**

---

🔹 How Decorators Work Internally

This line:

```python
@my_decorator
def show():
    print("Hello")
```

👉 Internally becomes:

```python
show = my_decorator(show)
```

So:

* `show` is passed to decorator
* decorator returns a new function (`wrapper`)
* now `show` = `wrapper`

---

🔹 Function as First-Class Object

Decorators work because Python treats functions as **first-class objects**

👉 That means:

* Functions can be assigned to variables
* Passed as arguments
* Returned from another function

```python
def greet():
    print("Hello")

x = greet   # function assigned
x()
```

---

 🔹 Closure Concept (Important 🔥)

A **closure** means:
👉 Inner function remembers variables of outer function

```python
def outer():
    def inner():
        print("Hello from inner")
    return inner

x = outer()
x()
```

✔ Decorators use this concept to keep reference of original function

---

 🔹 Structure of a Decorator

```python
def decorator(func):          # Step 1: Take function
    def wrapper(*args, **kwargs):   # Step 2: Wrapper
        # extra work
        result = func(*args, **kwargs)
        # extra work
        return result
    return wrapper            # Step 3: Return wrapper
```

 3. Multiple Decorators

```python

```

👉 Execution order:

```python
func = A(B(func))
```

---

 🔹 Decorator with Arguments (Advanced 🔥)

👉 When decorator itself takes parameters

```python
def repeat(n):
    def decorator(func):
        def wrapper():
            for i in range(n):
                func()
        return wrapper
    return decorator

@repeat(3)
def say_hi():
    print("Hi")

say_hi()
```

 🔹 Important Concepts to Remember

 1. `*args` and `**kwargs`

Used to handle any number of arguments

---

 2. Return Value Handling

```python
def decorator(func):
    def wrapper():

    return wrapper
```

---

 3. Metadata Issue (Advanced)

Decorator changes function name:

```python
print(func.__name__)  # may show wrapper
```

✔ Solution: use `functools.wraps`

```python
from functools import wraps

def decorator(func):
    @wraps(func)
    def wrapper():
        return func()
    return wrapper
```

````python
# 1. @staticmethod : which is called using classnname
# 👉 Does NOT use instance/object (self) or class (cls)

class MathUtils:
    @staticmethod
    def add(a, b):
        return a + b

print(MathUtils.add(5, 3))

# func call using object
# m=MathUtils()
# m.add()
````

````python
# 2. @classmethod
# 👉 Uses class (cls) instead of object (self)

class Student:
    # class variables 
    school_name = "ABC School"

    @classmethod
    def change_school(cls, name):
        # self.school_name=name
        cls.school_name = name

Student.change_school("XYZ School")
print(Student.school_name)
````

````python
# 3. @property
# 👉 Used to make a method behave like a variable

class Student:
    def __init__(self,marks):
        self._marks = marks

    @property
    def marks(self):
        return self._marks

s = Student(90)
print(s.marks)   # no parentheses

# print(s.marks())
````

````python
# 4. @setter (with property)
class Student:
    def __init__(self):
        self._marks = 0

    @property
    def marks(self):
        return self._marks # getter method

    @marks.setter
    def marks(self, value):
        if value < 0:
            print("Invalid marks")
        else:
            self._marks = value

s = Student()
s.marks = 85
print(s.marks)
````

````python
# 2. User-Defined Decorators
# 👉 You create these yourself to modify function behavior

# Example 1: Basic Decorator
def my_decorator(func): # outer function
    def wrapper(): # inner function
        print("Before function")
        func()
        print("After function")
    return wrapper

@my_decorator
def show():
    print("Hello")
    print("Welcome")

show()
````

````python
# Example 2: With Arguments

def decorator(func):
    def wrapper(name):
        print("Welcome")
        func(name)
    return wrapper

@decorator
def greet(name):
    print("Hello", name)

greet("Kartiki")
````

````python
# Example 3: Login Check (Real-Time)

def login_required(func):
    def wrapper(is_logged_in):
        if is_logged_in:
            func(is_logged_in)
        else:
            print("Access Denied")
    return wrapper

@login_required
def dashboard(is_logged_in):
    print("Welcome to Dashboard")

dashboard(True)
dashboard(False)
````

````python
import math

class Shape:
    def sh(self):
        self.sides = int(input("Enter number of sides: "))


class Rectangle(Shape):
    def rect(self):
        self.length = float(input("Enter length: "))
        self.breadth = float(input("Enter breadth: "))
        area = self.length * self.breadth
        print("Rectangle area =", area)


class Triangle(Shape):
    def tri(self):
        self.base = float(input("Enter base: "))
        self.height = float(input("Enter height: "))
        area = 0.5 * self.base * self.height
        print("Triangle area =", area)


class Circle(Shape):
    def cir(self):
        self.radius = float(input("Enter radius: "))
        area = math.pi * self.radius * self.radius
        # area = 3.14 * self.radius * self.radius
        print("Circle area =", area)


class Pentagon(Shape):
    def pen(self):
        self.side = float(input("Enter side length: "))
        area = (5 * self.side * self.side) / (4 * math.tan(math.pi/5))
        print("Pentagon area =", area)


# Objects
r = Rectangle()
r.sh()
r.rect()

t = Triangle()
t.sh()
t.tri()

c = Circle()
c.sh()
c.cir()

p = Pentagon()
p.sh()
p.pen()
````

# **`super()` Function in Python**

## **Definition**

The `super()` function is used to **call the methods or constructor (`__init__`) of the parent class** from the child class.

It helps avoid rewriting code and allows the child class to reuse the functionality of the parent class.

---

# **Advantages of `super()`**

* Reuses the parent class code.
* Reduces code duplication.
* Makes programs easier to maintain.
* Allows the child class to extend the functionality of the parent class.

---

# **Syntax**

```python
super().method_name()
```

or

```python
super().__init__()
```

---

# **Example 1: Calling Parent Class Method**

```python
class Person:

    def display(self):
        print("I am Person")

class Student(Person):

    def show(self):
        super().display()
        print("I am Student")

s = Student()
s.show()
```

### **Output**

```
I am Person
I am Student
```

### **Explanation**

* `Student` inherits `Person`.
* `super().display()` calls the `display()` method of the parent class.
* Then the child class prints `"I am Student"`.

---

# **Example 2: Using `super()` with Constructor**

```python
class Person:

    def __init__(self):
        self.name = input("Enter Name: ")

class Student(Person):

    def __init__(self):
        super().__init__()
        self.roll = int(input("Enter Roll Number: "))

    def display(self):
        print("Name:", self.name)
        print("Roll Number:", self.roll)

s = Student()
s.display()
```

### **Sample Output**

```
Enter Name: Rahul
Enter Roll Number: 101
Name: Rahul
Roll Number: 101
```

### **Explanation**

* `super().__init__()` calls the parent class constructor.
* The parent constructor accepts the student's name.
* The child constructor accepts the roll number.
* Both values are displayed using the child class method.

---

# **Example 3: Parent Constructor with Parameters**

```python
class Employee:

    def __init__(self, name):
        self.name = name

class Manager(Employee):

    def __init__(self, name, salary):
        super().__init__(name)
        self.salary = salary

    def display(self):
        print("Name:", self.name)
        print("Salary:", self.salary)

m = Manager("Amit", 50000)
m.display()
```

### **Output**

```
Name: Amit
Salary: 50000
```

# **Exam Theory (2 Marks)**

**Q. What is `super()` in Python?**

**Answer:**

The `super()` function is used to access the methods and constructor of the parent class from the child class. It helps in code reusability and avoids rewriting the parent class code.

**Example:**

```python
class A:
    def display(self):
        print("Class A")

class B(A):
    def show(self):
        super().display()
        print("Class B")

obj = B()
obj.show()
```

**Output**

```
Class A
Class B
```

---

## **Interview Question**

### **Q. Why do we use `super()` instead of directly calling the parent class?**

**Answer:**

* `super()` automatically refers to the parent class, making the code easier to maintain.
* If the parent class name changes, you do not need to update every method call.
* It is especially useful in **multiple inheritance**, where Python follows the **Method Resolution Order (MRO)** to determine which parent method to call.


````python
class Person:

    def display(self):
        print("I am Person")

class Student(Person):

    def show(self):
        super().display()
        print("I am Student")

s = Student()
# s.display()
s.show()
````

````python
class Person:

    def __init__(self):
        self.name = input("Enter Name: ")

class Student(Person):

    def __init__(self):
        # explicit call to parent constructor using super()
        super().__init__()
        self.roll = int(input("Enter Roll Number: "))

    def display(self):
        print("Name:", self.name)
        print("Roll Number:", self.roll)

s = Student()
s.display()
````

1. OOP (Object-Oriented Programming)

Definition:
OOP is a programming approach where we organize code using **classes and objects**.

Purpose of OOP:
* Makes code **easy to understand**
* Helps in **code reusability**
* Improves **security (data hiding)**
* Makes programs **modular and structured**
---

2. Class

Definition:
A **class** is a blueprint or template used to create objects.
It defines properties (variables) and methods (functions).

Example:
* class Student:
    name = "Kartiki"

---

3. Object

Definition:
An **object** is an instance of a class.
It represents a real-world entity and can use the properties and methods of the class.

Example:
* s1 = Student()
print(s1.name)
* s2=Student()
print(s2.name)
---

4. Constructor

Definition:
A **constructor** is a special method in a class that is automatically called when an object is created.

* In Python, it is written as `__init__()`

Example:

* class Student:
    def __init__(self, name):
        self.name = name

s1 = Student("Kartiki")
print(s1.name)


Output:

```
Kartiki
```

---

Short Summary (for revision):

1. **Class** → Blueprint
2. **Object** → Instance of class
3. **OOP** → Programming using classes & objects
4. **Constructor** → Auto-called method when object is created






````python
class Student:
    def show(self):
        print("This is a Student Class ")

    def display(self):
        print("I'm in Display CLass")


#object
s1=Student()
s1.show()
s1.display()

s2=Student()
s2.display()
````

````python
class Employee:
    id=1
    name="ram"
    designation="developer"
    dept="cs"


e=Employee()
print("Emp ID is :: ",e.id)
print("Emp Name is :: ",e.name)
print("Emp Designation is :: ",e.designation)
print("Emp Department is :: ",e.dept)

e1=Employee()
print("Emp ID is :: ",e1.id)
print("Emp Name is :: ",e1.name)
print("Emp Designation is :: ",e1.designation)
print("Emp Department is :: ",e1.dept)
````

````python
class Employee:
    def accept(self,name,salary):
        self.name=name
        self.salary=salary 

    def display(self):
        print("Name is :: ",self.name)
        print("Salary is :: ",self.salary)


e1=Employee()
e2=Employee()


e1.accept("Ram",50000)
e2.accept("Raj",30000)

e1.display()
e2.display()
````

````python
class Employee:
    def __init__(self,name,sal):
        self.name=name
        self.sal=sal

    def display(self):
        print("Name is :: ",self.name)
        print("Salary is :: ",self.sal)

e1=Employee("Ram",50000)
e2=Employee("Raj",30000)

e1.display()
e2.display()
````

````python
class Bank:
    def __init__(self,balance):
        self.balance=balance

    def display_bal(self):
        print("Current Balance is :: ",self.balance)


b=Bank(50000)
b.display_bal()
        
````

````python
#update value in object 

class Student:
    def __init__(self,name):
        self.name=name

s=Student("Amit")
print("Current value :: ",s.name)

s.name="Ram"
print("Updated value :: ",s.name)
````

````python
class EvenOdd:
    def check(self,num):
        if num%2==0:
            print("Even ")
        else:
            print("Odd")

n=EvenOdd()
n.check(8)
n1=EvenOdd()
n1.check(9)
````

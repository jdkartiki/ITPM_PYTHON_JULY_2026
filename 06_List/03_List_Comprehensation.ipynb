###### LIST COMPRENENSATION
* is a short and efficient way to create a list using single line statement  with loops and condition
* Syntax : new_list=[expression for item in iterable]

```python
print("hello")
```

```python
# sqaure of element in list 

lst=[]

n=int(input("Enter Count of Elements to Store :: "))
for i in range(n):
    num=int(input("Enter Elements :: "))
    lst.append(num)

print("Original list :: ",lst)

#using single line statement
square=[num*num for num in lst]

print("Sqaure of element in list :: ",square)
```

```python
# Cube of element in list
lst=[]

n=int(input("Enter Count of Elements to Store :: "))

for i in range(n):
    num=int(input("Enter Elements :: "))
    lst.append(num)

print("Original list :: ",lst)

cube=[num**3 for num in lst]

print("Cube of element in list :: ",cube)
```

```python
# even elements from list
lst=[]

n=int(input("Enter Count of Elements to Store :: "))

for i in range(n):
    num=int(input("Enter Elements :: "))
    lst.append(num)


print("Original list :: ",lst)

even=[num for num in lst if num%2==0]
print("Even elements in list :: ",even)
```

```python
# odd elements from list
lst=[]

n=int(input("Enter Count of Elements to Store :: "))

for i in range(n):
    num=int(input("Enter Elements :: "))
    lst.append(num)


print("Original list :: ",lst)

odd=[num for num in lst if num%2!=0]
print("odd elements in list :: ",odd)
```

```python
# Convert the int element from list to string 

lst=[]

n=int(input("Enter Count of Elements to Store :: "))

for i in range(n):
    num=int(input("Enter Elements :: "))
    lst.append(num)


print("Original list :: ",lst)

str_list=[str(num) for num in lst]
print("String element on list :: ",str_list)
print(type(lst))
print(type(str_list))
```

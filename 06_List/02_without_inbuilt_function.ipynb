```python
# find the max element from list 

lst=[]

n=int(input("Enter Count of Elements to Store :: "))

for i in range(n):
    num=int(input("Enter Elements :: "))
    lst.append(num)

print("Original list :: ",lst)

largest=lst[0]

for i in lst:
    if i>largest:
        largest=i

print("LArgest element is  :: ",largest)
```

```python
#remove the duplications from list 

lst=[]

n=int(input("Enter Count of Elements to Store :: "))

for i in range(n):
    num=int(input("Enter Elements :: "))
    lst.append(num)


print("Original list :: ",lst)

no_duplicates=[]
for i in lst:
    if i not in no_duplicates:
        no_duplicates.append(i)

print("List witout duplication :: ",no_duplicates)
```

```python
# reverse a list 

lst=[]

n=int(input("Enter Count of Elements to Store :: "))

for i in range(n):
    num=int(input("Enter Elements :: "))
    lst.append(num)


print("Original list :: ",lst)

rev=[]
for i in range(len(lst)-1,-1,-1):
    rev.append(lst[i])

print("Reverse List is :: ",rev)
```

```python
lst=[10, 20, 30, 40, 50]
i=len(lst)-1
rev=[]

while i>=0:
    rev.append(lst[i])
    i=i-1


print("Original list :: ",lst)
print("Reverse list :: ",rev)
```

```python
# separate the even and odd element from list

lst=[]

n=int(input("Enter Count of Elements to Store :: "))

for i in range(n):
    num=int(input("Enter Elements :: "))
    lst.append(num)


print("Original list :: ",lst)

even=[]
odd=[]

for i in lst:
    if i%2==0:
        even.append(i)
    else:
        odd.append(i)


print("Even element list :: ",even)
print("Odd element list :: ",odd)
```

```python
# even and  odd index 

lst=[]

n=int(input("Enter Count of Elements to Store :: "))

for i in range(n):
    num=int(input("Enter Elements :: "))
    lst.append(num)


print("Original list :: ",lst)

even_index=[]
odd_index=[]
for i in range(len(lst)):
    if i%2==0:
        even_index.append(lst[i])
    else:
        odd_index.append(lst[i])


print("Even Index element :: ",even_index)
print("Odd Index Element :: ",odd_index)
```

```python
# fetch duplicates 
lst=[]

n=int(input("Enter Count of Elements to Store :: "))

for i in range(n):
    num=int(input("Enter Elements :: "))
    lst.append(num)


print("Original list :: ",lst)

duplicates=[]
for i in lst:
    if lst.count(i)>1 and i not in duplicates:
        duplicates.append(i)

print("Dupliacte Element :: ",duplicates)
```

```python
# create 2D array matrix

matrix=[]
rows=int(input("Enter Number of Rows :: "))
cols=int(input("Enter Number of Columns :: "))

for i in range(rows):
    rows=[]
    for j in range(cols):
        val=int(input("Enter Value :: "))
        rows.append(val) 
    matrix.append(rows)

print("Matrix is :: ")
for r in matrix:
    for c in r:
        print(c,end=" ")
    print()
```

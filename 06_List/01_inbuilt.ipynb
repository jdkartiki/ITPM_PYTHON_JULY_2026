```python
# append : it is used to add elements from last index

lst=[]

n=int(input("Enter Count of Elements to Store :: "))

for i in range(n):
    num=int(input("Enter Elements :: "))
    lst.append(num)


print("Original list :: ",lst)

#append
element=int(input("Enter element to add :: "))
lst.append(element)

print("Updated list is :: ",lst)
```

```python
# extend- will allow you to add multiple elements at a time

lst1=[]
lst2=[]

n=int(input("Enter Count of Elements to Store in List 1 :: "))

for i in range(n):
    num=int(input("Enter Elements :: "))
    lst1.append(num)


n=int(input("Enter Count of Elements to Store in List 2 :: "))

for i in range(n):
    num=int(input("Enter Elements :: "))
    lst2.append(num)


print("List 1 :: ",lst1)
print("List 2 :: ",lst2)

print("=================================")

lst1.extend(lst2)
print("List 1 :: ",lst1)
print("List 2 :: ",lst2)
```

```python
# insert -- used to add element on specific index

lst=[]

n=int(input("Enter Count of Elements to Store :: "))

for i in range(n):
    num=int(input("Enter Elements :: "))
    lst.append(num)


print("Original list :: ",lst)

ind=int(input("enter Index Number :: "))
val=int(input("Enter Value to add :: "))

lst.insert(ind,val)
print("Updated List :: ",lst)
```

```python
# delete the value : remove()
lst=[]

n=int(input("Enter Count of Elements to Store :: "))

for i in range(n):
    num=int(input("Enter Elements :: "))
    lst.append(num)


print("Original list :: ",lst)

val=int(input("Enter Value to delete :: "))

if val in lst:
    lst.remove(val)
else:
    print("Value not found ")

print("Updated List :: ",lst)

# note :: remove() will take one element at a time to delete 
# delete first occurance of number 
```

```python
# delete the element using index  : pop

lst=[]

n=int(input("Enter Count of Elements to Store :: "))

for i in range(n):
    num=int(input("Enter Elements :: "))
    lst.append(num)


print("Original list :: ",lst)

ind=int(input("enter index to delete element :: "))

if ind<len(lst):
    lst.pop(ind)
else:
    print("Index Not found")

print("Updated list :: ",lst)
```

```python
# find index of element
lst=[]

n=int(input("Enter Count of Elements to Store :: "))

for i in range(n):
    num=int(input("Enter Elements :: "))
    lst.append(num)


print("Original list :: ",lst)

val=int(input("Enter Value to find index :: "))

if val in lst:
    print("Element is found at index :: ",lst.index(val))
else:
    print("Value not found")
```

```python
# find occurance of elements
lst=[]

n=int(input("Enter Count of Elements to Store :: "))

for i in range(n):
    num=int(input("Enter Elements :: "))
    lst.append(num)


print("Original list :: ",lst)

val=int(input("Enter element to count the frequency :: "))
print("Total Occurance of given element is :: ",lst.count(val))
```

```python
# sort the list 
lst=[]

n=int(input("Enter Count of Elements to Store :: "))

for i in range(n):
    num=int(input("Enter Elements :: "))
    lst.append(num)


print("Original list :: ",lst)

# ascending 
lst.sort()
print("Sorted List :: ",lst)

# descending
lst.sort(reverse=True)
print("Sorted List :: ",lst)
```

```python
# reverse the list 
lst=[]

n=int(input("Enter Count of Elements to Store :: "))

for i in range(n):
    num=int(input("Enter Elements :: "))
    lst.append(num)


print("Original list :: ",lst)

lst.reverse()
print("Reversed list :: ",lst)
```

```python
# copy the list 
lst=[]

n=int(input("Enter Count of Elements to Store :: "))

for i in range(n):
    num=int(input("Enter Elements :: "))
    lst.append(num)


print("Original list :: ",lst)
new_list=lst.copy()
print("Copied List :: ",new_list)
```

```python
# min,max,sum
lst=[]

n=int(input("Enter Count of Elements to Store :: "))

for i in range(n):
    num=int(input("Enter Elements :: "))
    lst.append(num)


print("Original list :: ",lst)

print("Minimum ELement in list :: ",min(lst))
print("Maximum ELement in list :: ",max(lst))
print("sum ELement in list :: ",sum(lst))
```

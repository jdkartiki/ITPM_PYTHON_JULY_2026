### IF-ELSE STATEMENT
* if - when the condition is true , will check the condition
* else - when the condition is false , no condition will be cjhecked 

```python
# postive and negative 
num=int(input("Enter number :: "))

if num>0:
    print("Positive Number ")
else:
    print("negative number")
```

```python
# loan eligibility - 50K

salary=float(input("enter salary :: "))
if salary>=50000:
    print("Eligible for Loan")
else:
    print("NOT eligible for loan ")
```

```python
# check the validity of recharge 
days=int(input("Enter Days :: "))

if days>0:
    print("Recharge is Active ")
else:
    print("Recharge is Expired!! Please Recharge ")
```

```python
# take the bill and give free delivery for >500

bill=float(input("Enter Bill amount :: "))

if bill>500:
    print("Free Delivery ")
else:
    print("Delivery Charges Are Applicable ")
```

* = is used to assign the value
    * 1. eg : num=10,name="Ram"

* == is used for comparision between operator/ variables with values
    * 1. eg : a=10 , b=10 -----> true 
    * 2. eg : a=10, b="10" -----> true

* === is used for comparision of values and datatype both 
     * 1. eg : a=10 , b=10 -----> true 
     * 2. eg : a=10, b="10" -----> False

```python
# username="admin" and password="admin@1234"

username=input("Enter Username :: ")
password=input("Enter Password :: ")

if username=='admin' and password=='admin@1234':
    print("Login Successful ")
else:
    print("Login Failed !! Please check Credentails ")
```

```python
# check the strength of password 

password=input("Enter Password :: ")

if len(password)>8:
    print("Strong Password ")
else:
    print("weak Password ")
```

```python
# check for weekend : sat and sun

day=input("Enter Dayname in lowercase :: ")

if day=='sat' or day=='sun':
    print("Its WEEKEND !!! ")
else:
    print("Its WEEKDAY ")
```

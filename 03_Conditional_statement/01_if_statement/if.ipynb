#### IF STATEMENT 
* If statement is used to check the condition
* if true will give the output
* syntax : if condition:

```python
print("hello")
```

```python
# check whether number is positive / negative

num=int(input("Enter number :: "))
if num>0:
    print("positive number ")

if num<0:
    print("negative number ")
```

```python
# whether num is even / odd

num=int(input("Enter number :: "))
if num%2==0:
    print("Even number ")

if num%2!=0:
    print("Odd Number ")
```

```python
# eligible for voting 
age=int(input("enter age :: "))
if age>=18:
    print("Eligible for Voting")

if age<18:
    print("NOT eligible for voting")
```

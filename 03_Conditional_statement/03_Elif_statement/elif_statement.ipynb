### ELIF STATEMENTS 
* having multiple statements - more than 2 conditions 
* will check mutliple conditions but when it is true will leave the next condition to be checked 

```python
# num is positive / negative / =0

num=int(input("Enter number :: "))

if num>0:
    print("Positive ")
elif num<0:
    print("Negative")
else:
    print("Number is equal to 0")
```

```python
# accept the percentage and on that display the grade 

percentage=float(input("enter the percentage :: "))

if percentage>=75 and percentage<=100:
    print("Grade A ")
elif percentage>=55 and percentage<75:
    print("Grade B")
elif percentage>=35 and percentage<55:
    print("Grade C")
else:
    print("Fail")
```

```python
# accept the colour of signal 

color=input("ENter colour of signal :: ")

if color=="red":
    print("STOP")
elif color=="green":
    print("GO")
elif color=="yellow":
    print("PLEASE SLOW THE VEHICLE ")
else:
    print("INVALID COLOR")
```

```python
# check the customer membership -- premium , gold, regular 

balance=int(input("Enter Balance :: "))

if balance>1000000:
    print("Premium Memebership Customer")
elif balance<1000000 and balance>=500000:
    print("Gold Memebership Customer ")
else:
    print("Regular Memebership Customer")
```

```python
# accept the bill and based on bill amt apply discount

bill=float(input("Enter Bill AMpunt :: "))

if bill>=10000:
    discount=bill*0.25
    bill=bill-discount
    print("Discount is :: ",discount)
    print("Total Bill AMount :: ",bill)
elif bill<10000 and bill>=5000:
    discount=bill*0.15
    bill=bill-discount
    print("Discount is :: ",discount)
    print("Total Bill AMount :: ",bill)
elif bill<5000 and bill>=2000:
    discount=bill*0.05
    bill=bill-discount
    print("Discount is :: ",discount)
    print("Total Bill AMount :: ",bill)   
else:
    print("NO DISCOUNT APPLICABLE !!!")
    print("Total Bill AMount :: ",bill)
```

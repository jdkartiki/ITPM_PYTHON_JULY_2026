```python
# creating login system 

username=input("Enter Username :: ")
password=input("Enter Password :: ")

if username=="admin":
    if password=="admin@1234":
        print("Login Successful ")
    else:
        print("Invalid Password")
else:
    print("Invalid Username ! Login Failed ")
```

```python
# accept the bill anf give the offer 

bill=float(input("Enter Bill Amount :: "))
prime_mem=input("Enter yes / no :: ")

if bill>10000:
    if prime_mem=="yes":
        print("Extra 10% Off ")
    else:
        print("5 % Off ")
else:
    print("No Discount !!! ")
```

```python
# accept salary and exp of emp and check bonus eligibility for emp 


sal=float(input("Enter Sal :: "))
exp=int(input("Enter Experience  in YEars :: "))

if sal>=50000:
    if exp>=5:
        print("Eligible for Bonus")
    else:
        print("Experience is less than 5 Years! Not eligible for bonus ")
else:
    print("Salary is less than 50000 ! Not eligible for bonus")
```
